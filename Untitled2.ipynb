```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('default')
```

```python
trades = pd.read_csv('historical_data.csv')
sentiment = pd.read_excel('fear_greed_index.xlsx')
```

```python
trades.head()
```

```python
sentiment.head()
```

```python
print(trades.columns)
print(sentiment.columns)
```

```python
trades.isnull().sum()
```

```python
sentiment.isnull().sum()
```

```python
trades['time'] = pd.to_datetime(trades['time'])
sentiment['date'] = pd.to_datetime(sentiment['date'])
```

```python
trades['date'] = trades['time'].dt.date
sentiment['date'] = sentiment['date'].dt.date
```

```python
merged = pd.merge(
    trades,
    sentiment,
    on='date',
    how='left'
)
```

```python
merged.groupby('classification')['Closed PnL'].mean()
```

```python
merged.groupby('classification').size()
```

```python
merged['win'] = merged['Closed PnL'] > 0
```

```python
merged.groupby('classification')['win'].mean()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(
    data=merged,
    x='classification',
    y='Closed PnL'
)
plt.title('Average Profit by Sentiment')
plt.show()
```

```python

plt.figure(figsize=(8,5))
sns.countplot(
    data=merged,
    x='classification'
)
plt.title('Trading Activity by Market Sentiment')
plt.show()
```

```python
merged['win'] = merged['Closed PnL'] > 0

win_rate = merged.groupby('classification')['win'].mean().reset_index()

plt.figure(figsize=(8,5))
sns.barplot(
    data=win_rate,
    x='classification',
    y='win'
)
plt.title('Win Rate by Market Sentiment')
plt.ylabel('Win Rate')
plt.show()
```

## Key Insights

1. Traders were most active during Greed periods, indicating increased market participation when sentiment was positive.

2. Average profit was highest during Extreme Fear periods, suggesting that contrarian trading strategies may be effective.

3. Win rates were lower during Extreme Greed despite higher trading activity, indicating possible overconfidence among traders.

4. Trading performance varied significantly across sentiment categories, showing that market psychology influences trader behavior.

5. Fear and Greed classifications can be used as valuable indicators for designing risk-management strategies.
